```python
#剪刀石頭布遊戲

import random

choice = ['a','b','c']
player = None
computer = None
game = 0
player_win = 0
computer_win = 0
check = {'a':'剪刀','b':'石頭','c':'布'}

while player_win !=2 and computer_win!=2:
    game += 1
    print('=============game',game,'===================')
    player = input('請出拳a.剪刀, b.石頭, c.布: ')
    while player not in choice:
        player = input('你只能出a.剪刀, b.石頭, c.布。再出一次:')
    computer = random.choice(['a','b','c'])
    print('你出"',check.get(player),'" vs 電腦出"',check.get(computer),'"')
    
    if computer == 'a':
        if player == 'a':
            print('平手')
        elif player == 'b':
            print('你獲勝!')
            player_win += 1
        elif player == 'c':
            print('電腦獲勝!')
            computer_win += 1
    
    if computer == 'b':
        if player == 'b':
           print('平手')
        elif player == 'c':
            print('你獲勝!')
            player_win += 1
        elif player == 'a':
            print('電腦獲勝!')
            computer_win += 1
    
    if computer == 'c':
        if player == 'c':
            print('平手')
        elif player == 'a':
            print('你獲勝!')
            player_win += 1
        elif player == 'b':
            print('電腦獲勝!')
            computer_win += 1

    print('你',player_win," - ",computer_win,'電腦')
if player_win ==2:
    print('#### 恭喜你獲得總冠軍 ####')
else:
    print('#### 恭喜電腦獲得總冠軍 ####')
```
