```python
import random

answer = random.randint(1,99) #產生1~99隨機整數為答案
guess_count = 0 #設定起始猜測次數為0
last_guess_max = 100
last_guess_min = 0

print('歡迎來到猜數字遊戲!!!')

while True: #無窮迴圈，直到猜對才 break
    guess = int(input('請猜一個0~100之間的整數: '))
    guess_count +=1
    if guess > answer:
        print('試試猜小一點 (答案在',last_guess_min,'~',guess,'之間)')
        last_guess_max = guess
    elif guess < answer:
        print('試試猜大一點 (答案在',guess,'~',last_guess_max,'之間)')
        last_guess_min = guess
    else:
        print('猜對了，正確答案是',answer)
        print('總共猜了',guess_count,'次，厲害唷')
        break
```
