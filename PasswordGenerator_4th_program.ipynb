```python
#強密碼產生
import random
#digit = list(range(5,21)) #密碼位數5~20之間。 Gemini不建議使用list(range)限制密碼位數，5 <= pw_digit <= 20更省記憶體
lower_case = ('abcdefghijklmnopqrstuvwxyz')
upper_case = ('ABCDEFGHIJKLMNOPQRSTUVWXYZ')
num = ('0123456789')
symbols = ('!@#$%^&*')
all_case = ('abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!@#$%^&*')

'''
以下是Fir寫的原版，無法反應輸入非數字字元
pw_digit = int(input('請問你需要幾位數的密碼？(最多20位數)'))
while pw_digit not in digit: #限制密碼位數5~20之間數字，否則重新輸入
    pw_digit = int(input('輸入錯誤，請問你需要幾位數的密碼？(最多20位數)'))
'''
#Gemini建議版本，使用 Python 的防護罩指令 —— try...except (嘗試...捕捉例外)。
#邏輯是： 「電腦你試著 (try) 把輸入轉成整數，如果成功就繼續；如果出錯 (except)，不要崩潰，印出警告就好
while True: 
    user_input = input('請問你需要幾位數的密碼？(5~20): ')
    try:
        # 1. 嘗試轉成整數
        pw_digit = int(user_input)
        
        # 2. 檢查範圍 (用這種數學寫法比 list(range) 更省記憶體喔！)
        if 5 <= pw_digit <= 20:
            break # 成功且範圍正確，跳出迴圈
        else:
            print("請輸入 5 到 20 之間的數字！")
            
    except ValueError:
        # 3. 如果轉整數失敗 (使用者輸入了英文或符號)
        print("輸入錯誤！請輸入「數字」就好。")


print('你的密碼是',pw_digit,'位數')
lc = random.choice(lower_case) #取樣1個lower_case，保證複雜度邏輯
uc = random.choice(upper_case) #取樣1個upper_case，保證複雜度邏輯
nm = random.choice(num) #取樣1個num，保證複雜度邏輯
sb = random.choice(symbols) #取樣1個symbols，保證複雜度邏輯
ac = random.choices(all_case,k = int(pw_digit) - 4) #隨機取樣補足密碼位數

#print(lc,uc,nm,sb,ac)

password = [lc,uc,nm,sb] + ac #結合前面取樣
#print(password)

random.shuffle(password) #打亂密碼順序
print(''.join(password)) #使用''.join把列表黏成字串

```
