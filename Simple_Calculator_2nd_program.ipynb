```python
# --- 1. 定義工具箱 (函式) ---

# 新增：專門用來美化數字顯示的函式
def format_number(n):
    if isinstance(n, float) and n.is_integer():
        return int(n)
    return n    
def add(x, y):
    return x + y
def subtract(x, y):
    return x - y     # 請在此完成減法
def multiply(x, y):
    return x * y    # 請在此完成乘法
def divide(x, y):
    if y == 0:
        return "無法計算 (除數不可為 0)" # 這裡直接回傳字串，讓主程式印出來更好看
    else:
        return x / y    # 記得加 return !
        
# --- 2. 主程式邏輯 ---
print("=== 簡單計算機 ===")
print("請選擇運算：")
print("1.相加")
print("2.相減")
print("3.相乘")
print("4.相除")

choice = input("請輸入選擇(1/2/3/4): ")
while choice not in ["1","2","3","4"]:
    print("輸入錯誤，請輸入 1-4 之間的選項")
    choice = input("請輸入選擇(1/2/3/4): ")

num1 = float(input("請輸入第一個數字: "))  # 提示：輸入的數字可能是小數，建議用 float()
num2 = float(input("請輸入第二個數字: "))

# 計算結果
result = 0 # 先初始化變數
if choice == '1':
    print(format_number(num1), "+", format_number(num2), "=", format_number(add(num1, num2)))
elif choice == '2':
    # 呼叫減法函式 
    print(format_number(num1), "-", format_number(num2), "=", format_number(subtract(num1, num2)))
elif choice == '3':
    # 呼叫乘法函式 
    print(format_number(num1), "*", format_number(num2), "=", format_number(multiply(num1, num2)))
elif choice == '4':
    # 呼叫除法函式
    print(format_number(num1), "/", format_number(num2), "=", format_number(divide(num1, num2)))
```
